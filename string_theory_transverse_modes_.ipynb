```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

#string parameters
L = np.pi
sigma = np.linspace(0, L, 300)

#superposition of two modes (n=1 and n=3)
def string_position(t):
    mode1 = 1.0 * np.cos(1 * t) * np.cos(1 * sigma)
    mode2 = 0.5 * np.cos(2 * t) * np.cos(2 * sigma)
    mode3 = 0.25 * np.cos(3 * t) * np.cos(3 * sigma)
    return mode1 + mode2 + mode3

fig, ax = plt.subplots(figsize=(8, 4))
ax.set_xlim(0, L)
ax.set_ylim(-2, 2)
ax.set_title("Relativistic Open String: Transverse Oscillations")
ax.set_xlabel("Sigma (coordinate along string)")
ax.set_ylabel("X (transverse displacement)")
ax.grid(True)

line, = ax.plot([], [], color="crimson", lw=2)

def init():
    line.set_data([], [])
    return line,

def update(frame):
    t = frame * 0.05
    X = string_position(t)
    line.set_data(sigma, X)
    return line,

ani = animation.FuncAnimation(fig, update, frames=100, init_func=init, interval=40, blit=True)
plt.close()
HTML(ani.to_jshtml())
```
